```python
# 문제 0 

# 1. 데이터 수집·전처리부터 모델 학습, 성능 평가까지 이어지는 머신러닝 파이프라인의 전체 구조를 명확히 제시하세요.
- 데이터 수잡과 전처리, 모델학습, 성능평가는 데이터 분석의 핵심이다. 먼저 각 단계별로 어떤 일을 수행하는지 간단히 정리해보고자한다.
    먼저 데이터 수집은 문제해결에 필요한 기초데이터를 수집하는 과정이다. 데이터의 크기보단 적합한 데이터를 확보하는것이 중요하다.
데이터 전처리는 정제되지 않는 일반적인 데이터를 모델이 학습할 수 있게 가공하는 것이다. 결측치와 이상치를 처리하고 스케일링과 인코딩을 주요작업으로한다
이 단계에서 데이터의 누수현상이 발생할 수 있다. 다음으로 성능평가는 분류와 회귀가 있는데 핵심은 과적합 여부를 판단하는 것이다.
    


# 2. 학습(Train) 및 평가(Test) 과정에서 발생할 수 있는 데이터 누수(Data Leakage)를 방지하기 위한 데이터 분리 및 처리 전략을 포함하여 기술하세요.
-> 대표적으로 데이터의 누수가 발생하는 이유는 표준화를 먼저하고, 데이터를 분할하기때문이다. 이는 강의 도중에도 자주 언급됐던 주의 사항중 하나이다.
이뿐만이 아니라, 데스트 데이터를 학습에 사용하는 것도 이유가된다. 이미 훈련을하며 학습했던 데이터이기에 정확도 높은 성능을 평가하기에 어려움이 있다.
종합적으로 데이터의 누수는 예측과정에서 사용해서는 안되는 정보가 포함되는 현상이다. 따라서 데이터 누수를 방식하기 위해서는 데이터를 먼저 분할한 후, 
훈련 데이터 기준으로 전처리를 해야한다.
```
